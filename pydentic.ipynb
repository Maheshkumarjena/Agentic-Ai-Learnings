<h1> Model validation </h1>

```python
from pydantic import BaseModel , EmailStr , AnyUrl, field_validator , model_validator
from typing import List, Dict , Optional 

class PatientData(BaseModel):
    name:str
    age:int
    linkedin_profile: Optional[AnyUrl] = None
    # we use EmailStr from pydantic to validate the email format
    email: Optional[EmailStr] = None

    @model_validator(mode='before')
    def validate_email_domain(cls, values):
        # model validator uses  multiple schema attributes  together to perform validation.
        email = values.get('email')
        if email is not None:
            valid_domains = ["gmail.com", "yahoo.com", "outlook.com"]
            domain_name = email.split("@")[-1]
            if domain_name not in valid_domains:
                raise ValueError("Invalid email domain")

        age = values.get("age")
        if age is not None:
            try:
                age = int(age)
            except (TypeError, ValueError):
                raise ValueError("Age must be an integer")
            if age < 0:
                raise ValueError("Age must be a positive integer")

            # Enforce a rule that links age with linkedin_profile presence.
            if values.get("linkedin_profile") is not None and age < 18:
                raise ValueError("LinkedIn profile is only allowed for patients 18 or older")

        return values


def add_patient_data(patient: PatientData):
    print(f"Patient Name: {patient.name}, Age: {patient.age}")
    print("Patient data added successfully.")
    pass

def update_patient_data(patient: PatientData):
    print(f"Patient Name: {patient.name}, Age: {patient.age}")
    print("Patient data updated successfully.")
    print(f"Updated Patient Name: {patient.name}, Email: {patient.email}, Age: {patient.age} , LinkedIn Profile: {patient.linkedin_profile}")
    pass

patient_data = { "name": "John Doe", "age": "30", "email": "johndoe@yahoo.com" , "linkedin_profile": "https://www.linkedin.com/in/johndoe" }

patient = PatientData(**patient_data)
add_patient_data(patient)
patient1 = PatientData(**{"name": "Jane Doe", "age": 22 , "linkedin_profile": "https://www.linkedin.com/in/janedoe", "email": "janedoe@gmail.com"})
update_patient_data(patient1)
```

```python
from pydantic import BaseModel , EmailStr , AnyUrl, field_validator , model_validator
from typing import List, Dict , Optional 

class PatientData(BaseModel):
    name:str
    height: float
    weight: float

    @property
    def bmi(self) -> float:
        return self.weight / (self.height ** 2)

    age:int
    linkedin_profile: Optional[AnyUrl] = None
    # we use EmailStr from pydantic to validate the email format
    email: Optional[EmailStr] = None

    @model_validator(mode='before')
    def validate_email_domain(cls, values):
        # model validator uses  multiple schema attributes  together to perform validation.
        email = values.get('email')
        if email is not None:
            valid_domains = ["gmail.com", "yahoo.com", "outlook.com"]
            domain_name = email.split("@")[-1]
            if domain_name not in valid_domains:
                raise ValueError("Invalid email domain")

        age = values.get("age")
        if age is not None:
            try:
                age = int(age)
            except (TypeError, ValueError):
                raise ValueError("Age must be an integer")
            if age < 0:
                raise ValueError("Age must be a positive integer")

            # Enforce a rule that links age with linkedin_profile presence.
            if values.get("linkedin_profile") is not None and age < 18:
                raise ValueError("LinkedIn profile is only allowed for patients 18 or older")

        return values


def add_patient_data(patient: PatientData):
    print(f"Patient Name: {patient.name}, Age: {patient.age} , Height: {patient.height}, Weight: {patient.weight}, BMI: {patient.bmi}")
    print("Patient data added successfully.")
    pass

def update_patient_data(patient: PatientData):
    print(f"Patient Name: {patient.name}, Age: {patient.age}")
    print("Patient data updated successfully.")
    print(f"Updated Patient Name: {patient.name}, Email: {patient.email}, Age: {patient.age} , LinkedIn Profile: {patient.linkedin_profile}")
    pass

patient_data = { "name": "John Doe", "age": "30", "email": "johndoe@yahoo.com" , "linkedin_profile": "https://www.linkedin.com/in/johndoe" , "height": 1.75, "weight": 70 }

patient = PatientData(**patient_data)
add_patient_data(patient)
patient1 = PatientData(**{"name": "Jane Doe", "age": 22 , "linkedin_profile": "https://www.linkedin.com/in/janedoe", "email": "janedoe@gmail.com", "height": 1.65, "weight": 60})
update_patient_data(patient1)
```

<h1>Nested Models </h1>

```python
class ContactInfo(BaseModel):
    phone: str
    email: EmailStr
    address: Optional[str] = None

class MedicalHistory(BaseModel):
    allergies: Optional[list[str]] = None
    medications: Optional[list[str]] = None
    past_conditions: Optional[list[str]] = None

class PatientData(BaseModel):
    name: str
    age: int
    contact_info: ContactInfo
    medical_history: MedicalHistory
    linkedin_profile: Optional[AnyUrl] = None

    @model_validator(mode='before')
    def validate_patient(cls, values):
        age = values.get("age")
        if age is not None:
            try:
                age = int(age)
            except (TypeError, ValueError):
                raise ValueError("Age must be an integer")
            if age < 0:
                raise ValueError("Age must be a positive integer")

        contact_info = values.get("contact_info")
        if isinstance(contact_info, dict):
            email = contact_info.get("email")
            if email is not None:
                valid_domains = ["gmail.com", "yahoo.com", "outlook.com"]
                domain_name = email.split("@")[-1]
                if domain_name not in valid_domains:
                    raise ValueError("Invalid contact email domain")

        if values.get("linkedin_profile") is not None and age is not None and age < 18:
            raise ValueError("LinkedIn profile is only allowed for patients 18 or older")

        return values

def add_patient_data(patient: PatientData):
    print(f"Patient Name: {patient.name}, Age: {patient.age}")
    print(f"Phone: {patient.contact_info.phone}, Email: {patient.contact_info.email}")
    print(f"Allergies: {patient.medical_history.allergies}")
    print("Patient data added successfully.")

def update_patient_data(patient: PatientData):
    print(f"Patient Name: {patient.name}, Age: {patient.age}")
    print(f"Address: {patient.contact_info.address}, Past Conditions: {patient.medical_history.past_conditions}")
    print("Patient data updated successfully.")

patient_data = {
    "name": "John Doe",
    "age": "30",
    "contact_info": {
        "phone": "123-456-7890",
        "email": "johndoe@yahoo.com",
        "address": "123 Main St"
    },
    "medical_history": {
        "allergies": ["peanuts", "shellfish"],
        "medications": ["aspirin"],
        "past_conditions": ["asthma"]
    },
    "linkedin_profile": "https://www.linkedin.com/in/johndoe"
}

patient = PatientData(**patient_data)
add_patient_data(patient)

patient1 = PatientData(
    name="Jane Doe",
    age=22,
    contact_info={
        "phone": "987-654-3210",
        "email": "janedoe@gmail.com",
        "address": "456 Elm St"
    },
    medical_history={
        "allergies": ["pollen"],
        "medications": None,
        "past_conditions": ["migraines"]
    },
    linkedin_profile="https://www.linkedin.com/in/janedoe"
)
update_patient_data(patient1)
```
